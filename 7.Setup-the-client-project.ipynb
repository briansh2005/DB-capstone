### Task 3

```python
join_query = """
SELECT 
    Customer.Name, 
    Customer.Contact, 
    Orders.TotalCost
FROM 
    Customer
INNER JOIN 
    Orders ON Customer.CustomerID = Orders.CustomerID
WHERE 
    Orders.TotalCost > 60;
"""

# 執行查詢
cursor.execute(join_query)

# 抓取所有查詢結果並列印出來
results = cursor.fetchall()
print("--- Customers with orders greater than $60 ---")
for row in results:
    print(f"Full Name: {row[0]} | Contact: {row[1]} | Total Cost: ${row[2]}")
```
